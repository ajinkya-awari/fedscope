```python
from datetime import datetime, timezone
from pathlib import Path
import json
import os
import re
import subprocess
import sys

WORK_ROOT = Path('/kaggle/working/fedscope_project')
TEST_ROOT = WORK_ROOT / 'tests'
EVIDENCE_ROOT = WORK_ROOT / 'evidence' / 'synthetic'
EVIDENCE_ROOT.mkdir(parents=True, exist_ok=True)
timestamp = datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%SZ')
evidence_path = EVIDENCE_ROOT / f'synthetic_validation_{timestamp}.json'
command = [sys.executable, '-B', '-m', 'pytest', '-q', '-p', 'no:cacheprovider', str(TEST_ROOT)]
print('STATUS: running only the documented offline synthetic validation command.')
print('COMMAND:', command)
result = subprocess.run(command, cwd=str(WORK_ROOT), env={**os.environ, 'PYTHONPATH': str(WORK_ROOT)}, capture_output=True, text=True, check=False)
passed_match = re.search(r'(\d+) passed', result.stdout)
failed_match = re.search(r'(\d+) failed', result.stdout)
payload = {
    'schema_version': 1, 'timestamp_utc': timestamp, 'status': 'passed' if result.returncode == 0 else 'failed',
    'exit_code': result.returncode, 'passed_count': int(passed_match.group(1)) if passed_match else None,
    'failed_count': int(failed_match.group(1)) if failed_match else None,
    'failure_category': None if result.returncode == 0 else 'synthetic_validation_command_failed',
    'command': command, 'evidence_policy': 'sanitized metadata; command output is not printed by this cell',
}
evidence_path.write_text(json.dumps(payload, indent=2, sort_keys=True) + '\n', encoding='utf-8')
print('EVIDENCE_PATH:', evidence_path)
if result.returncode != 0:
    raise SystemExit('STOP: synthetic validation failed; preserve the dated evidence JSON and do not continue.')
print('NEXT ACTION: inspect the sanitized evidence JSON in Cell 9.')
```

```python
from datetime import datetime, timezone
from pathlib import Path
import json
import subprocess
import sys

APPROVAL_COMMENT = ''  # Replace only after explicit approval for the bounded smoke command.
WORK_ROOT = Path('/kaggle/working/fedscope_project')
EVIDENCE_ROOT = WORK_ROOT / 'evidence' / 'live_smoke'
EVIDENCE_ROOT.mkdir(parents=True, exist_ok=True)
command = [sys.executable, str(WORK_ROOT / 'kaggle/run_fedscope.py'), '--mode', 'smoke', '--data-root', '/kaggle/working/medmnist', '--output-root', '/kaggle/working/fedscope_artifacts', '--device', 'auto']
print('LABEL: bounded federated training smoke; one round, seed 42, five clients.')
print('STATUS: final gated command is not executed automatically.')
if APPROVAL_COMMENT != 'I_APPROVE_FEDSCOPE_SMOKE':
    print('APPROVAL_REQUIRED: set the exact approval comment only after the preflight and synthetic evidence are reviewed.')
    print('COMMAND_NOT_RUN:', command)
    print('NEXT ACTION: obtain explicit approval for this bounded smoke command; do not start approved-seeds mode yet.')
    raise SystemExit(0)
timestamp = datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%SZ')
evidence_path = EVIDENCE_ROOT / f'bounded_smoke_{timestamp}.json'
try:
    result = subprocess.run(command, cwd=str(WORK_ROOT), capture_output=True, text=True, check=False, timeout=1800)
    status = 'passed' if result.returncode == 0 else 'failed'
    exit_code = result.returncode
except subprocess.TimeoutExpired:
    status, exit_code = 'timed_out', None
payload = {'schema_version': 1, 'timestamp_utc': timestamp, 'status': status, 'exit_code': exit_code, 'failure_category': None if status == 'passed' else 'bounded_smoke_failed_or_timed_out', 'command_label': 'bounded federated training smoke', 'evidence_policy': 'sanitized metadata; raw runner output is not printed'}
evidence_path.write_text(json.dumps(payload, indent=2, sort_keys=True) + '\n', encoding='utf-8')
print('STATUS:', status)
print('EVIDENCE_PATH:', evidence_path)
print('NEXT ACTION: stop and inspect sanitized smoke evidence; request separate approval before approved-seeds mode.')
```
